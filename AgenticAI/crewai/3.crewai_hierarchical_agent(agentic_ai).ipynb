### Hierarchical Multi-Agent (Agentic AI)
- Goal: Show manager agent delegating to sub-agents

```python
import os
from dotenv import load_dotenv
from crewai import Agent, Task, Crew, Process , LLM
load_dotenv()

llm = LLM(
    model="gemini/gemini-2.5-flash",
    api_key=os.getenv("GOOGLE_API_KEY"),
    temperature=0.3
)

# Manager / Orchestrator
manager = Agent(
    role="Project Manager",
    goal="Coordinate team to complete projects efficiently",
    backstory="You are an experienced PM who delegates effectively",
    allow_delegation=True,   # ← can delegate to others
    llm=llm
)

# Specialists
researcher = Agent(
    role="Researcher",
    goal="Provide thorough research",
    backstory="Expert researcher",
    llm=llm
)

analyst = Agent(
    role="Data Analyst",
    goal="Analyze data and provide insights",
    backstory="Expert data analyst",
    llm=llm
)

writer = Agent(
    role="Report Writer",
    goal="Write comprehensive reports",
    backstory="Expert report writer",
    llm=llm
)

# Only give ONE high level task to manager
main_task = Task(
    description="""Create a comprehensive market analysis report 
                   for {industry} industry. Include research, 
                   analysis and written report.""",
    expected_output="Complete market analysis report",
    agent=manager   # ← manager receives the goal
)

crew = Crew(
    agents=[researcher, analyst, writer], # we are not adding manager here, because manager is orchestrating, not executing tasks
    tasks=[main_task],
    process=Process.hierarchical,
     manager_agent=manager  # ← manager decides who does what
)


result = crew.kickoff(inputs={"industry": "Electric Vehicles"})
print(result)
```

### Multi-Agent Workflow
- UseCase:
  - A content creation pipeline with two agents — a researcher and a writer. 
  - The researcher will first research a given topic and extract key points, 
  - then the writer will use those key points to write an engaging blog post. 
  - This demonstrates how to create multiple agents, assign them specific tasks, 
  - and have them work together in a sequential process.

```python
import os
from dotenv import load_dotenv
from crewai import Agent, Task, Crew, Process , LLM
load_dotenv()

llm = LLM(
    model="gemini/gemini-2.5-flash",
    api_key=os.getenv("GOOGLE_API_KEY"),
    temperature=0.3
)

# Agent 1 — Researcher
researcher = Agent(
    role="Researcher",
    goal="Research topics thoroughly",
    backstory="You are an expert researcher with eye for detail",
    llm=llm
)

# Agent 2 — Writer
writer = Agent(
    role="Content Writer",
    goal="Write engaging content based on research",
    backstory="You are a skilled writer who makes content interesting",
    llm=llm
)

# Task 1 — Research (goes to researcher)
research_task = Task(
    description="Research about {topic} and find key points",
    expected_output="5 key research points about the topic",
    agent=researcher
)

# Task 2 — Write (goes to writer, uses research output)
write_task = Task(
    description="Write a short blog post using the research provided",
    expected_output="A 3 paragraph engaging blog post",
    agent=writer,
    context=[research_task]   # ← writer gets researcher's output
)

crew = Crew(
    agents=[researcher, writer],
    tasks=[research_task, write_task],
    process=Process.sequential,  # ← one after another
)

result = crew.kickoff(inputs={"topic": "Benefits of AI in Healthcare"})
print(result)
```

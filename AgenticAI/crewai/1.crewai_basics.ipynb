### Example to create an Agent

```python
import os
from dotenv import load_dotenv
from crewai import Agent, Task, Crew, LLM

load_dotenv()  # Load environment variables from .env file

llm = LLM(
    model = "gemini/gemini-2.5-pro",
    api_key = os.getenv("GOOGLE_API_KEY"),
    temperature = 0

)

# Create an agent
agent = Agent(
    # Defines who the agent is and what they do
    role = "Explainer",
    # defines the agent's overall purpose
    goal = "Explain topics in simple language",
    # Context to help the agent understand how to achieve the goal
    backstory = "You are a teacher who explains complex things simply",
    llm = llm

)

# Give it a task
task = Task(
    # defines what needs to be done
    description = "Explain what is AI in 3 lines",
    # Defines what the output should look like
    expected_output = "A 3 line simple explanation",
    agent = agent
)

# Run the task
crew = Crew(agents = [agent], tasks = [task])
result = crew.kickoff()
print(result)
```

### Agent with Dynamic Input

```python
import os
from dotenv import load_dotenv
from crewai import Agent, Task, Crew, LLM

load_dotenv()  # Load environment variables from .env file

llm = LLM(
    model = "gemini/gemini-2.5-pro",
    api_key = os.getenv("GOOGLE_API_KEY"),
    temperature = 0

)

# Define Agent

agent = Agent(
    role = "Summarizer",
    goal = "Summarizer any given text clearly",
    backstory = "You are an expert at creating conscise summaries of complex topics",
    llm = llm
)


# Define Task
task = Task(
    description = "Summarize the following topic: {topic}",
    expected_output = "A concise summary of the topic",
    agent = agent
)

# Run the task
crew = Crew(agents = [agent], tasks = [task])
result = crew.kickoff(inputs={"topic": "What is GenAI?"})
print(result)
```

### AI Agent Tool Calling
- Usecase:
   - A pizza shop assistant who can check pizza availability and show menu to customers.
   - We will create two tools - one for checking the delivery and another for fetching menu.
   - The agent will use these tools to help customers.

```python
import os
from dotenv import load_dotenv
from crewai import Agent, Task, Crew, LLM
from crewai.tools import tool # Importing tool decorator

load_dotenv()  # Load environment variables from .env file

llm = LLM(
    model = "gemini/gemini-2.5-pro",
    api_key = os.getenv("GOOGLE_API_KEY"),
    temperature = 0

)

# Tool1 - Normal function with @tool decorator to make it a tool that can be used by agents

@tool("Delivery Checker")
def check_delivery(pincode: str) -> str:
    "checks if delivery is available for a given pincode"
    available_pincodes = ["12345", "67890", "54321"]
    if pincode in available_pincodes:
        return f"Delivery is available for pincode {pincode}."
    else:
        return f"Sorry, delivery is not available for pincode {pincode}."

# Tool2 - Another normal function with @tool decorator
@tool("Menu Fetcher")
def get_menu(category: str) -> str:
    "Gets pizza menu items for a given category for veg or non-veg"
    menu = {
        "veg" : ["Margherita - $5", "Farmhouse - $7", "Peppy Paneer - $8"],
        "non-veg" : ["Pepperoni - $6", "Chicken Supreme - $8", "Meat Lovers - $10"]
    }
    items = menu.get(category.lower())
    if items:
        return f"Available {category} pizzas: " + ", ".join(items)
    else:
        return f"category {category} not found."


# Agent with both tools
agent = Agent(
    role = "Pizza order assistant",
    goal = "Help customers check delivery availability and show pizza menu",
    backstory = "You are a helpful pizza shop assistant at Domino's pizza shop.",
    tools = [check_delivery, get_menu], # passing both tools to the agent
    llm = llm
)    


task = Task(
    description = """
                    customer request:
                    Their pincode is {pincode}.
                    They are interested in {category} pizzas.
                    Check delivery availability and show the menu
                  """,
    expected_output = "Delivery status and list of available pizzas in the category",
    agent = agent
)


crew = Crew(agents = [agent], tasks = [task])
result = crew.kickoff(inputs={"pincode": "12345", "category": "veg"})
print(result)
```
